2) **Layer - > silver**
- Data Cleaning & validation for each column's cell values.

```python
df.dropna()
```

```python
## Check how many missing values exist per column
df.isna().sum()
```

#### -> Looks like the format of the cell values and columns has matched the pattern, so there's no need to change the format of a cell value.
Except for these columns: 
- cost_before_optimization_usd,
- cost_savings_usd,
- cost_savings_percent,
- resource_utilization_percent.

```python
df['cost_before_optimization_usd'] = df['cost_before_optimization_usd'].map('${:,.2f}'.format)
```

```python
df['cost_savings_usd'] = df['cost_savings_usd'].map('${:,.2f}'.format)
```

```python
df['cost_savings_percent'] = df['cost_savings_percent'].map('{:,.2f}%'.format)
```

```python
df['resource_utilization_percent'] = df['resource_utilization_percent'].map('{:,.2f}%'.format)
```

```python
import seaborn as sns
```

```python
df_1 = pd.read_csv('cloud_platform.csv')
```

```python
#Bar Chart
plt.figure(figsize=(10, 6))
provider_savings = df_1.groupby('cloud_provider')['cost_savings_usd'].mean().sort_values(ascending = False)
sns.barplot(
    x = provider_savings.index,
    y = provider_savings.values
)
plt.title("Average cost savings by cloud provider",
          fontsize = 15 )
plt.xlabel("Cloud Provider")
plt.ylabel(" Average Cost Savings (USD)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("plot1_cost_savings.png", dpi = 300)
plt.show()
```

```python
##Scatter plot
plt.figure(figsize = (10,6))
sns.scatterplot(
    data = df_1,
    x = 'compute_hours',
    y = 'cost_after_optimization_usd',
    hue = 'cloud_provider',
    alpha = 0.6
)

plt.title("Compute hours Vs cost After Optimization", fontsize=15)
plt.xlabel("Compute Hours")
plt.ylabel("cost after optimization (USD)")
plt.legend(title = 'cloud provider', bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.savefig('plot2_compute_cost.png', dpi = 300)
plt.show()
```

```python
## correlation heatmap
numeric_cols = [
    'compute_hours',
    'vcpus',
    'storage_gb',
    'memory_gb',
    'usage_units',
    'job_runs']
correlation = df[numeric_cols].corr()
plt.figure(figsize= (12,9))
sns.heatmap(
    correlation,
    annot = True,
    cmap = 'coolwarm',
    fmt = '.2f',
    linewidths = 0.5)
plt.title("Correlation Heatmap of cloud Usage Variables",fontsize = 15)
plt.tight_layout()
plt.savefig('plot3_correlation_heatmap.png', dpi =300)
plt.show()
```
